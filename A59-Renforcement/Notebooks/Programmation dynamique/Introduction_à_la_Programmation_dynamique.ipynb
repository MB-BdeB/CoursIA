# Programmation dynamique

La **programmation dynamique** est une technique de résolution de problèmes qui divise un problème complexe en sous-problèmes plus simples. Cette méthode consiste à mémoriser les solutions des sous-problèmes pour éviter de les recalculer plusieurs fois, ce qui permet de gagner du temps et d’améliorer l'efficacité de l'algorithme. C'est particulièrement utile pour des problèmes où des sous-problèmes se répètent, comme dans le calcul de suites ou de chemins optimaux.

### Principe de base

Le principe fondamental de la programmation dynamique repose sur deux concepts principaux :

1.  **Sous-problèmes qui se répètent** : Le problème peut être divisé en plusieurs sous-problèmes identiques ou similaires. Par exemple, dans la suite de Fibonacci, calculer :

    $$F(n) = F(n-1) + F(n-2)$$

    nécessite de calculer plusieurs fois les valeurs de $F(n-1)$ et $F(n-2)$.

2.  **Mémoire pour éviter les recalculs** : On enregistre (ou "mémorise") les résultats des sous-problèmes déjà calculés, pour qu’ils soient immédiatement disponibles quand ils sont à nouveau nécessaires. Ce processus est souvent appelé **mémoïsation**.

### Types de programmation dynamique

Il y a deux approches courantes pour implémenter la programmation dynamique :

-   **Approche par le haut (Top-down)** : Dans cette approche, le problème est résolu en divisant le problème principal en sous-problèmes et en résolvant les sous-problèmes de manière récursive. Les résultats des sous-problèmes sont stockés dans une table (ou cache) pour éviter les calculs répétés. C'est ce qu'on appelle la **mémoïsation**.

-   **Approche par le bas (Bottom-up)** : Dans cette approche, on résout les sous-problèmes les plus petits d’abord et on utilise ces résultats pour construire des solutions aux sous-problèmes de taille plus grande, jusqu’à ce que le problème complet soit résolu. Cette approche utilise souvent des boucles pour remplir un tableau de résultats sans récursivité.

### Exemple simple : Suite de Fibonacci

Calculer la suite de Fibonacci est un bon exemple pour montrer l’efficacité de la programmation dynamique. La suite est définie par la relation :

$$F(n) = F(n-1) + F(n-2)$$

avec les conditions initiales :

$$F(0) = 0 \quad \text{et} \quad F(1) = 1$$

#### Solution naïve (sans programmation dynamique)

En utilisant une simple fonction récursive, on appelle à plusieurs reprises la fonction pour les mêmes valeurs, ce qui entraîne beaucoup de calculs redondants.

#### Solution avec programmation dynamique

Avec la programmation dynamique, on stocke les valeurs de $F(n-1)$ et $F(n-2)$ dans un tableau et on les utilise directement si elles ont déjà été calculées. Cela réduit significativement le nombre de calculs, améliorant l'efficacité de l'algorithme de manière exponentielle à linéaire.

### Applications courantes

La programmation dynamique est largement utilisée pour résoudre des problèmes d'optimisation où l’on doit trouver la meilleure solution parmi plusieurs, comme :

-   **Problèmes de chemins optimaux** : Par exemple, pour trouver le chemin le plus court dans un graphe, la relation de récurrence pour la distance minimale $d(i, j)$ entre deux points $i$ et $j$ est souvent formulée comme :

    $$d(i, j) = \min(d(i, k) + d(k, j)) \quad \text{pour tous } k$$

-   **Problème du sac à dos (knapsack)** : Soit $v_i$ la valeur et $w_i$ le poids de l'objet $i$. Si la capacité du sac est $W$, la solution optimale est donnée par :

    $$V(W) = \max(V(W - w_i) + v_i)$$

Ces équations démontrent l’efficacité de la **programmation dynamique** dans la résolution de problèmes d’optimisation en réduisant les calculs redondants.

# Lien entre la programmation dynamique et les techniques d'apprentissage par renforcement

### Programmation dynamique et apprentissage par renforcement

La programmation dynamique est un cadre général pour résoudre des problèmes de décision séquentiels, où l’on cherche une séquence d'actions qui maximise une récompense cumulative. En apprentissage par renforcement, on utilise ces techniques pour optimiser les actions d'un agent dans un environnement, afin de maximiser la récompense obtenue sur le long terme.

#### Q-learning et TD Learning

1.  **Q-learning** : C'est une méthode de **Temporal Difference (TD) Learning**. Elle met à jour une fonction de valeur d'état-action $Q(s, a)$ en utilisant une approximation de la valeur future pour chaque état-action. Q-learning utilise une formule inspirée de la programmation dynamique pour estimer les valeurs des actions possibles dans chaque état, sans nécessiter un modèle complet de l'environnement.

    La mise à jour de $Q(s, a)$ dans Q-learning suit cette équation :
    $$Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right]$$
    Ici, on ajuste la valeur de $Q(s, a)$ en fonction de la récompense immédiate $r$ et de la valeur de l'état suivant $s'$, en utilisant un **apprentissage par différence temporelle (TD)**. Le terme $$\max_{a'} Q(s', a')$$ représente l'estimation de la meilleure récompense possible depuis l'état suivant $s'$.

2.  **TD Learning** : C'est une famille de méthodes d'apprentissage qui inclut Q-learning et SARSA. Les méthodes TD combinent la **mise à jour incrémentale** (comme en programmation dynamique) avec l’apprentissage direct basé sur l’expérience, en utilisant des valeurs estimées de futurs états sans attendre que les épisodes soient terminés.

3.  **Equations de Bellman** : Les équations de Bellman servent de base théorique pour la programmation dynamique dans les MDP (Processus de Décision Markovien). Elles définissent la relation de récursivité pour la valeur d’un état ou d’une action, en décomposant la récompense totale attendue en une récompense immédiate et une récompense future.

    L’équation de Bellman pour la valeur de l’état $V(s)$ est donnée par :
    $$V(s) = \max_{a} \left[ r(s, a) + \gamma \sum_{s'} P(s' | s, a) V(s') \right]$$
    et pour la valeur de l'état-action $Q(s, a)$, elle est :
    $$Q(s, a) = r(s, a) + \gamma \sum_{s'} P(s' | s, a) \max_{a'} Q(s', a')$$
    Ces équations montrent comment chaque valeur dépend des valeurs futures (principe de récursivité), un aspect essentiel en programmation dynamique.

### Pourquoi ces méthodes relèvent de la programmation dynamique

Ces méthodes relèvent de la programmation dynamique parce qu’elles reposent sur des **équations récursives** pour estimer les valeurs d'états ou d’actions en fonction des valeurs futures, tout en **mémorisant** (ou en réutilisant) ces valeurs pour éviter de recalculer les mêmes sous-problèmes.

En somme, Q-learning, TD Learning, et les équations de Bellman utilisent les principes de la programmation dynamique pour résoudre les problèmes de décision séquentielle en maximisant les récompenses cumulées.
