```python
import numpy as np
import matplotlib.pyplot as plt

def estimate_expected_value(num_throws):
    sum_of_dice_values = 0
    running_average = []

    for i in range(num_throws):
        dice_value = np.random.randint(1, 7)
        sum_of_dice_values += dice_value
        running_average.append(sum_of_dice_values / (i + 1.0))

    average = sum_of_dice_values / num_throws
    print(f"Valeur moyenne après {num_throws} lancers: {average}")

    plt.plot(running_average, label="Moyenne empirique")
    plt.axhline(y=3.5, color='k', linestyle='-', label="Valeur attendue théorique (3.5)")
    plt.legend()
    plt.show()

estimate_expected_value(100000)
```

# Explication du code

-------------------

Dans le code ci-haut, on utilise un programme pour estimer la moyenne d'un grand nombre de lancers de dé, et on compare cette moyenne (appelée "moyenne empirique") à la "valeur attendue théorique". Voici une explication des deux concepts pour t’aider à mieux comprendre ce qui se passe dans le code.

### 1. Valeur attendue théorique (3.5)

La **valeur attendue théorique** d'un lancer de dé est la moyenne calculée en fonction des probabilités et des valeurs possibles d'un lancer de dé à six faces. Avec un dé équilibré, chaque face a une probabilité de \( \frac{1}{6} \), et la valeur attendue théorique se calcule comme :

$$
E(X) = 1 \cdot \frac{1}{6} + 2 \cdot \frac{1}{6} + 3 \cdot \frac{1}{6} + 4 \cdot \frac{1}{6} + 5 \cdot \frac{1}{6} + 6 \cdot \frac{1}{6} = 3.5
$$

Cette valeur théorique de 3.5 est celle qu'on obtiendrait **en moyenne** si on lançait un dé un nombre infini de fois.

### 2. Moyenne empirique

La **moyenne empirique** est la moyenne calculée à partir des résultats obtenus en lançant effectivement le dé un certain nombre de fois (par exemple, 10 000 fois). Comme le dé est lancé de manière aléatoire, la moyenne empirique varie à chaque exécution du programme et dépend des valeurs tirées.

Dans le code, la variable `running_average` conserve la moyenne cumulative des lancers au fur et à mesure qu'ils sont effectués. Par exemple, après 1 lancer, la moyenne empirique est juste le résultat de ce lancer. Après 10 lancers, la moyenne empirique est la somme des 10 résultats divisée par 10, et ainsi de suite.

### Explication du code

- **Initialisation** : La variable `sum_of_dice_values` garde la somme des valeurs des lancers. La liste `running_average` stocke la moyenne après chaque lancer, pour suivre comment elle se rapproche de la valeur attendue théorique de 3.5.
- **Boucle** :
  - À chaque itération, une valeur de dé (`dice_value`) est générée aléatoirement entre 1 et 6.
  - Cette valeur est ajoutée à `sum_of_dice_values`.
  - La moyenne cumulative (`sum_of_dice_values / (i + 1)`) est calculée et ajoutée à `running_average`.
- **Affichage** :
  - Après la boucle, `average` affiche la moyenne finale après tous les lancers.
  - Le graphique montre comment la moyenne empirique (en bleu) se stabilise autour de 3.5 (la ligne noire), la valeur attendue théorique, à mesure que le nombre de lancers augmente.

En exécutant ce code, tu verras que la **moyenne empirique** se rapproche de la **valeur attendue théorique** (3.5) quand le nombre de lancers devient grand.
