# Introduction à la Méthode Monte Carlo en Apprentissage par Renforcement

---

### Objectif de la Méthode Monte Carlo
La méthode **Monte Carlo** est une technique d’estimation dans l'apprentissage par renforcement, utilisée pour **estimer les valeurs des actions** et **améliorer les politiques** en fonction des résultats observés dans un environnement.

#### Points Clés :
- **Utilisation d'épisodes complets** : Monte Carlo utilise l'ensemble des expériences d'un épisode (début à fin) pour calculer les valeurs.
- **Approche sans modèle** : Contrairement aux méthodes basées sur un modèle, Monte Carlo n’a pas besoin de connaître les transitions d’état (probabilités de passage d’un état à l'autre).

---

### Principe de Base
La méthode Monte Carlo repose sur l'idée de **moyennes d'échantillons** pour estimer les valeurs. Cela signifie qu’elle **calcule la valeur d'une action** en prenant la moyenne des récompenses observées pour cette action sur plusieurs épisodes.

### Formule d'Estimation
L'objectif est de **calculer l'espérance** des retours pour un état ou une action en fonction des épisodes joués.

#### Estimation de la Valeur d'un État \(V(s)\)
La valeur de l'état \( s \) est calculée comme la moyenne des retours \( G_t \) pour tous les épisodes où l'état \( s \) a été visité.

$$
V(s) = \frac{1}{N(s)} \sum_{i=1}^{N(s)} G_t
$$

où :
- **\( V(s) \)** : Valeur estimée de l'état \( s \).
- **\( N(s) \)** : Nombre de fois que l'état \( s \) a été visité.
- **\( G_t \)** : Retour total obtenu à partir du moment où \( s \) est visité jusqu’à la fin de l'épisode.

#### Exemple de Calcul
Imaginons que vous soyez à l'état \( s \) et que vous ayez observé les retours suivants lors de différents épisodes : \(10\), \(20\), et \(15\). Si \( s \) a été visité 3 fois, alors :

$$
V(s) = \frac{10 + 20 + 15}{3} = 15
$$

---

### Types de Monte Carlo

Il existe deux variantes principales pour estimer les valeurs avec Monte Carlo :

1. **Méthode de la Première Visite**
   - Cette méthode ne prend en compte que la **première fois** où l'état \( s \) est visité dans chaque épisode.
   - Elle ignore les visites répétées de \( s \) dans le même épisode.

   > *Exemple* : Si \( s \) est visité deux fois dans le même épisode, seul le retour de la première visite est pris en compte pour le calcul de \( V(s) \).

2. **Méthode de Toutes les Visites**
   - Dans cette méthode, **chaque visite** de l'état \( s \) est prise en compte dans chaque épisode.
   - Ainsi, si \( s \) est visité plusieurs fois dans le même épisode, chaque visite contribue à la moyenne.

   > *Exemple* : Si \( s \) est visité deux fois dans un épisode, les deux retours seront utilisés pour calculer \( V(s) \).

---

### Mise à Jour avec la Formulation **(1 - alpha)**

On peut également reformuler la mise à jour de Monte Carlo pour conserver une part de la valeur actuelle en utilisant **(1 - alpha)**. Cela permet de montrer comment la nouvelle estimation est pondérée entre l'ancienne estimation et la nouvelle information.

#### Équation de Mise à Jour avec **(1 - alpha)**

Pour chaque état **s**, la valeur **V(s)** peut être mise à jour ainsi :

$$
V(s) \leftarrow (1 - \alpha) V(s) + \alpha G_t
$$





### Équation de Mise à Jour avec \(1 - alpha\)

La mise à jour de la valeur  **V(s)** en utilisant une pondération entre l'ancienne et la nouvelle estimation peut s'écrire comme suit :

$$
V(s) \leftarrow (1 - \alpha) V(s) + \alpha G_t
$$

où :

- $$ (1 - \alpha) V(s) $$ : Part de la **valeur actuelle** conservée.
- $$ \alpha G_t $$ : Nouvelle estimation basée sur le retour total \( G_t \).

---

### Explication

- $$ (1 - \alpha) V(s) $$ : Cette partie pondère la **valeur actuelle** de **s**, en lui donnant une importance réduite par  **(1 - alpha)**.
- $$ \alpha G_t $$ : Cette partie ajoute une pondération **alpha** au **nouveau retour** observé $$ G_t $$, ajustant ainsi **V(s)** en fonction des nouvelles informations.





---

### Pourquoi Utiliser Monte Carlo ?

Monte Carlo est particulièrement utile dans des situations où :
- **L’environnement est inconnu** : Il n'est pas nécessaire d’avoir un modèle des transitions d’états.
- **Les récompenses sont aléatoires** : La méthode moyenne les retours pour obtenir une estimation plus fiable.
- **La politique peut être améliorée** : À chaque épisode, on peut mettre à jour la politique pour maximiser les retours futurs.

En résumé, la méthode Monte Carlo est un outil puissant pour estimer des valeurs d'état et des valeurs d'action dans des environnements sans modèle, en se basant sur des observations.
