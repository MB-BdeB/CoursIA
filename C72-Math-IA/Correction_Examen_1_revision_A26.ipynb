## Question 1 — Loi normale

On a $X\sim\mathcal N(7,1{,}2^2)$ : moyenne $\mu=7$ ans et écart-type $\sigma=1{,}2$ an.

On veut limiter les retours à $2{,}5\%$. Un retour correspond à $X\le k$, donc :

$$
P(X\le k)=0{,}025.
$$

Le quantile correspondant est $z=-1{,}96$. En standardisant :

$$
\frac{k-7}{1{,}2}=-1{,}96.
$$

Ainsi :

$$
k=7-1{,}96(1{,}2)=4{,}648\approx4{,}6\text{ ans}.
$$

**Réponse : a) $k=4{,}6$ ans.**

La garantie doit être placée au 2,5e percentile de la durée de vie.

## Question 2 — Loi binomiale et moment d'ordre 2

Une face paire est l'une des faces $2,4,6$, donc :

$$
p=P(\text{paire})=\frac{3}{6}=\frac12.
$$

Sur 100 lancers indépendants, le nombre $X$ de faces paires suit donc :

$$
X\sim\mathcal B(100,\tfrac12).
$$

La première affirmation est **vraie**.

Pour une binomiale :

$$
E(X)=np=100\left(\frac12\right)=50,
$$
$$
\operatorname{Var}(X)=np(1-p)=100\left(\frac12\right)\left(\frac12\right)=25.
$$

On utilise l'identité :

$$
E(X^2)=\operatorname{Var}(X)+[E(X)]^2.
$$

Donc :

$$
E(X^2)=25+50^2=2525.
$$

**Réponses : vrai et $E(X^2)=2525$.**

## Question 3 — Variable aléatoire continue

La densité est :

$$
f(x)=
\begin{cases}
kx,&0\le x<2,\\
0,&\text{sinon.}
\end{cases}
$$

### 3.1 Fonction de répartition

Par définition :

$$
F(x)=P(X\le x)=\int_{-\infty}^{x}f(t)\,dt.
$$

Pour $0<x<2$, la densité est nulle avant 0, donc :

$$
F(x)=\int_0^x f(t)\,dt.
$$

**Réponse : b).**

### 3.2 Condition de densité

Une densité doit être positive et avoir une aire totale égale à 1 :

$$
f(x)\ge0,
$$
$$
\int_0^2f(x)\,dx=1.
$$

**Réponse : c).**

### 3.3 Calcul de $k$

$$
\int_0^2kx\,dx=1
\quad\Longrightarrow\quad
k\left[\frac{x^2}{2}\right]_0^2=1
\quad\Longrightarrow\quad
2k=1.
$$

Donc :

$$
k=\frac12.
$$
$$
\boxed{k=\frac12}
$$

**Réponse : c).**

### 3.4 Centrage d'une variable

Si $\mu=E(X)$, alors $\mu$ est une constante :

$$
E(X-\mu)=E(X)-E(\mu)=\mu-\mu=0.
$$

**Réponse : b).**

## Question 4 — Machines, probabilités totales et Bayes

Notons $D$ l'événement « la pièce est défectueuse ».

| Machine | $P(M)$ | $P(D\mid M)$ |
|---|---:|---:|
| A | $0{,}60$ | $0{,}01$ |
| B | $0{,}30$ | $0{,}02$ |
| C | $0{,}10$ | $0{,}05$ |

### 4.1 Machine A et pièce défectueuse

$$
P(A\cap D)=P(A)P(D\mid A)=0{,}60(0{,}01)=0{,}006=0{,}6\%.
$$

**Réponse : a).**

### 4.2 Pièce non défectueuse

Loi des probabilités totales :

$$
\begin{aligned}
P(D)&=P(A)P(D\mid A)+P(B)P(D\mid B)+P(C)P(D\mid C)\\
&=0{,}60(0{,}01)+0{,}30(0{,}02)+0{,}10(0{,}05)\\
&=0{,}017.
\end{aligned}
$$

Donc :

$$
P(D^c)=1-P(D)=1-0{,}017=0{,}983=98{,}3\%.
$$

**Réponse : d).**

### 4.3 Origine A sachant que la pièce est défectueuse

Formule de Bayes :

$$
P(A\mid D)=\frac{P(A\cap D)}{P(D)}=\frac{0{,}006}{0{,}017}=\frac6{17}\approx0{,}3529.
$$

**Réponse : c) $\frac6{17}$.**

## Question 5 — Tirage sans remise

L'urne contient 2 boules rouges, 2 vertes et 1 bleue. On tire 2 boules sans remise. $X$ compte les rouges et $Y$ compte les vertes.

### 5.1 Support de $X$

On peut obtenir 0, 1 ou 2 boules rouges :

$$
\operatorname{Supp}(X)=\{0,1,2\}.
$$

**Réponse : c).**

### 5.2 Lois de $X$ et $Y$

Il y a le même nombre de boules rouges et vertes. Les rôles rouge et vert sont symétriques, donc $X$ et $Y$ ont la même loi.

**Réponse : vrai.**

### 5.3 Tirer la boule bleue

Il y a $\binom52=10$ paires possibles. Pour contenir la boule bleue, on choisit l'une des 4 autres boules : $\binom41=4$ possibilités.

$$
P(\text{bleue tirée})=\frac{\binom41}{\binom52}=\frac4{10}=\frac25.
$$

**Réponse : b).**

### 5.4 Une rouge et une verte

$$
P(X=1,Y=1)=\frac{\binom21\binom21}{\binom52}=\frac4{10}=\frac25.
$$

**Réponse : b).**

### 5.5 Exactement une verte

On choisit une verte parmi 2 et une non-verte parmi 3 :

$$
P(Y=1)=\frac{\binom21\binom31}{\binom52}=\frac6{10}=\frac35.
$$

**Réponse : b).**

### 5.6 Indépendance

$$
P(X=1,Y=1)=\frac25,
$$
$$
P(X=1)P(Y=1)=\frac35\times\frac35=\frac9{25}.
$$

Comme $\frac25=\frac{10}{25}\ne\frac9{25}$, les événements ne sont pas indépendants.

**Réponse : faux.** Le tirage sans remise crée une dépendance entre les résultats.

## Question 6 — Intervalle de confiance et test sur une moyenne

Données :

$$
n=25,\qquad \bar x=5{,}6,\qquad s=3{,}2.
$$

L'écart-type de la population n'est pas connu. On utilise donc Student avec $n-1=24$ degrés de liberté.

### 6.1 Intervalle de confiance à 98 %

La formule est :

$$
\bar x\pm t_{\alpha/2;24}\frac{s}{\sqrt n}.
$$

Ici, $\alpha=0{,}02$, donc $t_{0,01;24}\approx2{,}492$.

Erreur standard :

$$
SE=\frac{3{,}2}{\sqrt{25}}=0{,}64.
$$

Marge d'erreur :

$$
ME=2{,}492(0{,}64)\approx1{,}595.
$$

Intervalle :

$$
5{,}6\pm1{,}595=[4{,}005,7{,}195]\approx[4{,}01,7{,}20].
$$

**Réponse : a) $[4{,}01;7{,}20]$.**

Interprétation : la méthode produit des intervalles qui contiennent la vraie moyenne dans environ 98 % des échantillons répétés.

### 6.2 Test de l'affirmation du fabricant

Le fabricant affirme que la moyenne ne dépasse pas 5 jours :

$$
H_0:\mu\le5,
$$
$$
H_1:\mu>5.
$$

Statistique :

$$
t=\frac{\bar x-5}{s/\sqrt n}=\frac{5{,}6-5}{0{,}64}=0{,}9375.
$$

Au seuil de 2 %, la valeur critique est environ $2{,}064$. Comme $0{,}9375<2{,}064$, on ne rejette pas $H_0$.

**Réponse : oui**, on ne dispose pas d'assez de preuves pour contredire l'affirmation du fabricant.

## Question 7 — Test sur une proportion

On observe $x=6$ animaux infectés sur $n=220$.

$$
\hat p=\frac6{220}\approx0{,}02727.
$$

La question est de savoir si la proportion est inférieure à 3 % :

$$
H_0:p=0{,}03,
$$
$$
H_1:p<0{,}03.
$$

Statistique de test :

$$
z=\frac{\hat p-p_0}{\sqrt{p_0(1-p_0)/n}}
=\frac{0{,}02727-0{,}03}{\sqrt{0{,}03(0{,}97)/220}}
+    \approx-0{,}237.
$$

Pour un test à gauche au seuil de 5 %, la valeur critique est $-1{,}645$. Comme :

$$
-0{,}237>-1{,}645,
$$
on ne rejette pas $H_0$.

**Réponse : non.** Les données ne permettent pas de conclure que la proportion réelle est inférieure à 3 %.

> Ne pas rejeter $H_0$ ne signifie pas que $H_0$ est prouvée vraie. Cela signifie que les preuves sont insuffisantes pour la rejeter.
