**Test d'indépendance**

Dans le but de mieux comprendre la relation entre la colère et les maladies cardiaques, on a recueilli un échantillon de 8 474 personnes ayant une pression artérielle normale et ne souffrant pas de maladie cardiaque et on les a suivis pendant quatre ans. Au début de l'étude, les sujets ont tous passé le test du niveau de colère de Spielberg qui mesure jusqu'à quel point une personne peut facilement se mettre en colère. Les sujets sont ensuite classifiés comme étant faiblement, modérément ou grandement colériques. À la fin de l'étude, ils sont classifiés comme souffrant de maladie cardiaque (crise cardiaque ou traitement médical pour une maladie cardiaque) ou ne souffrant pas de maladie cardiaque.

Voici le tableau du nombre de personnes dans chaque catégorie :

| | Faible | Modéré | Grand | Total |
| :--- | :---: | :---: | :---: | :---: |
| **Maladie cardiaque** | 53 | 110 | 27 | **190** |
| **Pas de maladie cardiaque** | 3 057 | 4 621 | 606 | **8 284** |
| **Total** | **3 110** | **4 731** | **633** | **8 474** |


**Questions :**
* **a)** Décrivez en mots les hypothèses d'intérêt dans ce problème.
* **b)** Calculez les fréquences théoriques (en les ajoutant au tableau ci-haut), et la statistique observée $\chi^2$ du test de *khi-deux*.
* **c)** Faites un test de *Khi-deux* de niveau $5\%$ pour les hypothèses en **a)** en respectant toutes les étapes. Quelles sont vos conclusions ? Y a-t-il un lien entre le niveau de colère et le fait d'avoir une maladie cardiaque ? Si oui, comment le décririez-vous ?


**Réponse**
* **a)**

$H_0$: Le niveau de colère et la maldie cardiaque sont indépendants.

$H_1$: Le niveau de colère et la maldie cardiaque sont dépendants.

* **b)** La statistique du test d'indépendance est
$$\chi^2=\sum_{i=1}^6 \frac{(O_i-E_i)^2}{E_i}=16,08\;.$$

* **c)**

**Validation du test**

• L'échantillon est aléatoire et suffisamment grand $n$
=8474>30.

• Toutes les fréquences théoriques calculées en b) sont largement supérieures à 5. **Le test est valide**.

Dans la table de la loi du Khi-deux, pour $\alpha=0,05$ et un degré de liberté de $(2-1)(3-1)=2$, la valeur critique $\chi_{0,05, 2}=5,99$.

**Décision:**
Puisque $\chi^2=16,08> 5,99$, nous rejetons largement l'hypothèse d'indépendance. Le niveau de colère et la maladie cardiaque sont significativement liés. Au niveau de signification de 5%, le niveau de colère et la maladie cardiaque semblent être dépendants.
