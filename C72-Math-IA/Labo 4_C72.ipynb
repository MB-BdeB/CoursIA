# Estimation par Intervalles de confiances:
**Test SAT d'admission à l'université**
Le pointage typique à la partie mathématique de cet examen est habituellement
de 475. La distribution du pointage est normale avec un écart type de
100. Mais comme ce n'est qu'une minorité des finissants du secondaire
qui font l'examen, quelqu'un fait le commentaire suivant: "Si tous les
finissants du secondaire faisaient l'examen, la moyenne ne dépasserait pas
450". Afin de vérifier cette hypothèse, on fait une expérience. On prend
un EAS de 500 finissants de la Californie et la moyenne de leur pointage
est de $\overline{x} = 461$.

1.  Donnez un intervalle de    confiance de niveau $95\%$ pour la moyenne du Test SAT en spécifiant  l'erreur standard et la marge d'erreur pour un niveau de confiance de  $95\%$ .

2. Rapportez l'intervalle de confiance de niveau  $95\%$ dans une phrase qui explique la signification du  $95\%$.

3. Quelle taille d'échantillon aurait-il fallu pour avoir une marge d'erreur de $5$ points ou moins ?




**Réponse**

1)

Un intervalle de confiance de niveau 95% est donné par
$$
[𝑥̅ -z_{0,025}\frac{\sigma}{\sqrt n}\;,\;𝑥̅ +z_{0,025}\frac{\sigma}{\sqrt n}] = [461 -1,96\frac{100}{\sqrt {500}}\;,\;461 +1,96\frac{100}{\sqrt {500}}]$$$$= [461 − 8,77\;,\; 461 + 8,77]
= [452,23; 469,77]$$
Avec $n= 500$,  la marge d’erreur de niveau 95%
était de 8,77 points.

2. Nous sommes confiants à 95% que la moyenne du Test SAT se situe entre 452,23 et
469,77 points, c'est-à-dire 461 ± 8,77 points. Autrement dit, ou bien la moyenne du Test SAT se situe entre 452,23 et
469,77 , ou bien non (lorsque nous somme dans les $5\%$ des cas des échantillons dont les intervalles de confiances ne contiennent pas la moyenne).

**Remarque**
$95\%$ est environ la probabilité qu'un échantillon aléatoire fournisse un intervalle de confiance contenant la moyenne de la poulation. Si notre échantillon en fait partie, la moyenne se situe entre 452,23 et  469,77.

```python
import math
from scipy import stats

# Données de l'énoncé
mu_0 = 450         # Moyenne sous l'hypothèse nulle (H0)
sigma = 100        # Écart-type de la population
n = 500            # Taille de l'échantillon
x_bar = 461        # Moyenne observée
alpha = 0.05
conf_level = 1 - alpha
z_critique = stats.norm.ppf(1-alpha/2)
print(f"Valeur critique= {z_critique:.2f}")


# 1. Calcul de l'erreur standard (SE)
error_standard = sigma / math.sqrt(n)
# Marge d'erreur (ME)
ME=z_critique*error_standard
print(f"Marge d'erreur (ME) = {ME:.2f}")
# Calcul d'un intervalle de confiance bilatéral à 95% pour illustrer
CI = [x_bar - z_critique * error_standard, x_bar + z_critique * error_standard]
print(f" Un intervalle de confiance de niveau {conf_level*100:.2f}% est [{CI[0]:.2f}, {CI[1]:.2f}]")
```


# Test d'hypothèse unilatéral sur une moyenne:
**Test SAT d'admission à l'université.**

 1. Testez l'hypothèse que la moyenne du Test SAT est de 450 contre l'hypothèse *unilatérale*. Assurez vous de bien spécifier les hypothèses nulle et alternative, et de calculer la statistique de test  ainsi que la valeur-p. Que conluez-vous ? Y a-t-il suffisamment d'évidence pour rejeter l'hypothèse
que la moyenne ne dépasse pas 450 ?

2.  Est-ce que le résultat en 1) est significatif  au niveau 1\%, l'est-il au niveau 0,5\% ?

**Réponse**
1. Nous confrontons les hypothèses

$H_0:$ $\mu=450$ (ou encore $\mu \leq 450$)

vs

$H_0:$ $\mu>450$ (observée sur l'échantillon).

On calcule la statistique du test unilatéral
$$z=\frac{\bar{x}-450}{\sigma/\sqrt n}=2,46\;.$$

**Décision: ** Comparaison de la statistique du test avec la valeur critique $z_{\alpha}=1,64$.

Ici, $z=2,46> z_{5\%}=1,64$, donc nous rejettons l'hypothèse nulle $H_0$. **Le résultat est significatif pour un niveau de $5\%$**.







```python
import math
from scipy import stats

# Données
mu_0 = 450
sigma = 100
n = 500
x_bar = 461

# Calcul de l'erreur standard et de la statistique z
error_standard = sigma / math.sqrt(n)
#statistique du test unilatéral
z_stat = (x_bar - mu_0) / error_standard

print(f"Statistique de test calculée z = {z_stat:.2f}\n")

# Boucle sur les différents seuils de signification alpha
for alpha in [0.05, 0.01, 0.005]:
    # Calcul du z critique (unilatéral à droite)
    z_alpha = stats.norm.ppf(1 - alpha)

    print(f"Pour alpha = {alpha} (valeur critique z_alpha = {z_alpha:.2f}) :")

    # Règle de décision : si z_stat > z_alpha, on rejette H0
    if z_stat > z_alpha:
        print("  Decision: On rejette l'hypothese nulle H_0.")
    else:
        print("  Decision: On ne peut pas rejeter l'hypothese nulle H_0.")
    print()
```
